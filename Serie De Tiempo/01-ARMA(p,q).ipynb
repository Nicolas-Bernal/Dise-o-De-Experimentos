## Ejemplo 2.2

```r
y1 = arima.sim(n = 100, list(ar = c(.8)))
y2 = arima.sim(n = 100, list(ar = c(-.7)))
y3 = arima.sim(n = 100, list(ar = c(1.2,-.9)))
y4 = arima.sim(n = 100, list(ma = c(.6)))
y5 = arima.sim(n = 100, list(ma = c(-.7)))
y6 = arima.sim(n = 100, list(ar = c(.8),ma = c(-.4)))

par(mfrow = c(3,2))

ts.plot(y1,col = 1,main = "y1")
ts.plot(y2,col = 2,main = "y2")
ts.plot(y3,col = 3,main = "y3")
ts.plot(y4,col = 4,main = "y4")
ts.plot(y5,col = 5,main = "y5")
ts.plot(y6,col = 6,main = "y6")
```

```r
par(mfrow = c(3,2))

acf(y1,col = 1)
acf(y2,col = 2)
acf(y3,col = 3)
acf(y4,col = 4)
acf(y5,col = 5)
acf(y6,col = 6)
```

```r
par(mfrow = c(3,2))

pacf(y1,col = 1)
pacf(y2,col = 2)
pacf(y3,col = 3)
pacf(y4,col = 4)
pacf(y5,col = 5)
pacf(y6,col = 6)
```

### Observaciones del ts.plot, acf, pacf

* Para y1 en el ts.plot no encontramos a simple vista una gran tendencia o estacionalidad marcada luego al miarar el acf encontramos que existe correlacion en los datos pero hasta el momento tampoco son estacionales ya que de caen de una forma suave mas luego en el pacf encontramos que totalmente no es estacional ya que ningun valor siquiera supera la banda de confianza generada por lo que el modelo y1 = arima.sim(n = 100, list(ar = c(.8))) no es estacional.

* Para y2 en el ts.plot encontramos una tendencia media de los datos la serie se comporta un poco estacional pero en el acf no encontramos autocorrelacion marcada en los datos por lo que empezamos a ver que no es estacional luego en el pacf encontramos que tampoco es estacional ya que no supera la banda de confianza.

* Para y3 en el ts.plot si encontramos una estacionalidad algo marcada que lo confirma el acf ya que muestra autocorrelacion y ademas una estacionalidad en la serie para algunos valores en el pacf encontramos que los datos si son estacionales nose salen de la banda de confianz apero continuan con la tendencia

* Para y4 en el ts.plot encontramos que no tiene estacionalidad lo que se corrobora con el acf el cual muestra que los datos no son correlacionados luego en el pacf se ve igual la no correlacion de los datos

* Para el y5 en el ts.plot encontramos que si tiene estacionalidad luego esto lo correbora el pacf ya que el acf no nos muestra una autocorrelacion directamente

* Para el y6 en el ts.plot encontramos una estacionalidad marcada lo cual lo confirma el acf y el pacf

# Ejemplo 3

Tomando la base de datos (adjunto) del Índice de Precios al Consumidor (IPC) de España. Aplicar transformaciones y diferenciaciones necesarias. Obtener gráficas de la serie original y transformadas-diferenciadas, igualmente los ACF, y los PACF tanto de las series originales y de la  series transformadas-diferenciadas. Interpretar.

```r
datos = read.csv2("/content/CPIEurope2000-15.csv")

IPC = datos[,9]

datos_tr = log(IPC)

datos_diff = diff(datos_tr)

par(mfrow = c(2,2))

acf(IPC,col = 1, main = "IPC")
acf(datos_diff,col = 2, main = "IPC-Transfomada")
pacf(IPC,col = 1, main = "IPC")
pacf(datos_diff,col = 2, main = "IPC-Transfomada")
```

### Observaciones

* Para la serie de tiempo sin transformar llega a hacer muy complicado intendar observar y dar informacion sobre eso mientras que aplicando una transformacion logaritmica y ademas una diferencia podemos conseguir que su estudio sea mas adecuado como se muentra en sus graficos ya transformados como en el acf se observa correlacion y ademas estacionalidad cosa que refuerza en pacf
